```python
# Q17
result = df.groupby('product').agg(
    total_units=('units_sold', 'sum'),
    avg_discount=('discount_pct', 'mean'),
    max_rating=('customer_rating', 'max'),
    min_delivery=('delivery_minutes', 'min')
)
print(result)
```

```python
# Q18
g = df.groupby('branch')
comparison = pd.DataFrame({
    'total_records_size': g.size(),
    'rating_count': g['customer_rating'].count()
})
comparison['missing_ratings'] = comparison['total_records_size'] - comparison['rating_count']
print(comparison)
print("size() counts all the rows in each group including the rows with missing ratings, but count() counts only the non-missing values of that column. So rating_count is less than or equal to size, and the difference is the number of missing ratings.")
```

```python
# Q19
def range_func(x):
    return x.max() - x.min()

print(df.groupby('product')['units_sold'].agg(range_func))
```

```python
# Q23
df_h = df.set_index(['branch', 'product'])
print(df_h.head())
print(df_h.groupby(level='branch')['units_sold'].sum())
```

```python
# Q24
pivot1 = pd.pivot_table(df, values='units_sold', index='branch', columns='product',
                        aggfunc='sum', margins=True, margins_name='Total')
print(pivot1)
```

```python
# Q25
pivot2 = pd.pivot_table(df, values=['discount_pct', 'customer_rating'], index='branch',
                        columns='product', aggfunc='mean')
print(pivot2)
```

```python
# Q26
df26 = df.copy()
missing_idx = df26.index[df26['customer_rating'].isnull()]
print("Before replacement:")
print(df26.loc[missing_idx, ['record_id', 'branch', 'product', 'customer_rating']])
print("Missing ratings:", df26['customer_rating'].isnull().sum())
df26['customer_rating'] = df26['customer_rating'].fillna(
    df26.groupby(['branch', 'product'])['customer_rating'].transform('mean')
)
print("After replacement:")
print(df26.loc[missing_idx, ['record_id', 'branch', 'product', 'customer_rating']])
print("Missing ratings:", df26['customer_rating'].isnull().sum())
```

```python
# Q27
def weighted_avg(g):
    return (g['units_sold'] * g['discount_pct']).sum() / g['units_sold'].sum()

print(df.groupby('branch')[['units_sold', 'discount_pct']].apply(weighted_avg))
```

```python
# Q28
top3 = df.groupby('branch')[['product', 'day', 'units_sold']].apply(lambda x: x.nlargest(3, 'units_sold'))
top3 = top3.reset_index(level=1, drop=True).reset_index()
print(top3)
```

```python
# Q29
df['sales_contribution_pct'] = (df['units_sold'] / df.groupby('branch')['units_sold'].transform('sum') * 100).round(2)
print(df[['record_id', 'branch', 'units_sold', 'sales_contribution_pct']].head(10))
print(df.groupby('branch')['sales_contribution_pct'].sum().round(2))
```

```python
# Q30
df['sales_category'] = pd.qcut(df['units_sold'], 4,
                               labels=['Low Sales', 'Below Average Sales', 'Above Average Sales', 'High Sales'])
summary = df.groupby('sales_category', observed=True).agg(
    total_units=('units_sold', 'sum'),
    avg_discount=('discount_pct', 'mean'),
    avg_rating=('customer_rating', 'mean'),
    avg_delivery=('delivery_minutes', 'mean'),
    num_records=('record_id', 'count')
)
print(summary)
```
